```python
import os
from glob import glob
from pathlib import Path

import geopandas as gpd
import pandas as pd
```

```python
# Load files
files = glob("/data/uscuni-restricted/01_preprocessed_census/*")

# Load file with relevant column information
names = pd.read_csv("column_names_geodem.csv", delimiter=";")

# Columns you always want to keep
common_columns = [
    "NUTS_2",
    "naz_oblast",
    "NUTS_3",
    "naz_kraj",
    "kod_okres",
    "naz_okres",
    "kod_orp",
    "naz_orp",
    "kod_obec",
    "naz_obec",
    "kod_mco",
    "nazev_mco",
    "nadzsjd",
    "nuts_2_regs",
    "nazev_regionu_soudrznosti",
    "nuts3_kraj",
    "nazev_kraje",
    "nazev_okresu",
    "nazev_orp",
    "nazev_obec",
]

# Determine columns to drop: keep those with use==1 OR in common_columns
drop = names[names["use"] != 1]

# Remove unnecessary columns column
drop = drop.drop(columns={"use"})

# Exclude common_columns from drop list
drop_cols = [c for c in drop["Column Names"] if c not in common_columns]

# Process all files
for file in files:
    path = Path(file)
    # Skip helper files that start with "_"
    if path.stem.startswith("_"):
        continue
    # Load and clean
    data = pd.read_csv(path)
    data_cleaned = data.drop(columns=drop_cols, errors="ignore")
    # Save the cleaned data
    data_cleaned.to_csv(
        f"/data/uscuni-restricted/geodem/{path.stem}_cleaned.csv",
        index=False,
    )
```

```python
# Read polygons
polygons = gpd.read_file(
    "/data/uscuni-restricted/Data_SLDB_2021/d_zsj_010122", driver="ESRI Shapefile"
)
polygons.head(2)
```

```python
# Prepare data for merging
# polygons = polygons.iloc[:, [0, 1, -1]]
polygons["geometry"] = polygons.geometry.force_2d()
polygons.head(2)
```

```python
# Read data for converting between zsj and nadzsj
converter = pd.read_excel(
    "/data/uscuni-restricted/Data_SLDB_2021/zsj_data/ZSJD_slouceneZSJD/sloucene/_prevodnik_zsjd_nadzsjd.xlsx"
)
```

```python
# Prepare data for merging
converter.columns = converter.iloc[2]
converter = converter.iloc[4:]
converter = converter.iloc[:, 18:]
# converter["kod_nadzsj_d"] = converter["kod_nadzsj_d"].astype(int)
converter.head(5)
```

```python
mapping = converter.set_index("kod_zsj_d")["kod_nadzsj_d"]
```

```python
polygons["kod_nadzsj_d"] = polygons["KOD_ZSJ_D"].map(mapping)
```

```python
polygons["kod_nadzsj_d"].isna().sum()
```

```python
polygons[polygons["kod_nadzsj_d"].isna()]
```

```python
polygons.loc[158, "kod_nadzsj_d"] = "53806003"
```

```python
nadzjs_polygons = polygons.dissolve("kod_nadzsj_d")
```

```python
files = glob("/data/uscuni-restricted/geodem/*nadzsj*")
files = [f for f in files if "cleaned" not in f]
files = [f for f in files if ".parquet" not in f]


len(files)
```

```python
# Process all files
for file in files:
    # Read path of the files
    path = Path(file)
    # Open data
    data = pd.read_csv(path, dtype={"nadzsjd": str})
    # Merge data
    data_df = pd.merge(
        data,
        nadzjs_polygons[["geometry"]],
        left_on="nadzsjd",
        right_index=True,
        how="left",
    )
    # Convert to gdf
    data_gdf = gpd.GeoDataFrame(data_df, geometry="geometry", crs=polygons.crs)
    data_gdf = data_gdf.dropna(subset="geometry")

    # Save the new dataset
    data_gdf.set_index("nadzsjd").to_parquet(
        f"/data/uscuni-restricted/geodem/{path.stem}.parquet"
    )
```

```python
# Create empty list to store all files
dfs = []

# Process all files
for file in files:
    dfs.append(gpd.read_parquet(file))
```

```python
dfs = [df.set_geometry(df.geometry) for df in dfs]

# Drop additional geometry columns before concatenation
for i in range(1, len(dfs)):
    dfs[i] = dfs[i].drop(columns=["geometry"])

# Concatenate along columns
concat_df = pd.concat(dfs, axis=1)

# Reassign geometry column after concatenation
concat_df = gpd.GeoDataFrame(concat_df, geometry=dfs[0].geometry)
```
