# Bandwidth vizualisation 

Plot a bandwidth curve for each class

```python
import os
import pathlib
from glob import glob

import geopandas as gpd
import joblib
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

from gwlearn.ensemble import GWRandomForestClassifier
from gwlearn.linear_model import GWLogisticRegression
from gwlearn.search import BandwidthSearch
```

```python
# Open scores for adaptive bandwidth
p_files = glob("/data/uscuni-restricted/06_bandwidths/*.csv")
p_files = [f for f in p_files if "fa" in f]
# p_files = [f for f in p_files if "/3" in f]
p_files = [f for f in p_files if "vs" in f]

p_files.sort()
```

```python
# bandwidth_dict = {#1: 950,
#      2: 200,
#      3: 950,
#      4: 1400,
#      5: 1850,
#        6: 800,
#       7: 2300,
#      8: 1850}
```

```python
plt.figure(figsize=(10, 10))

for file in p_files:
    df = pd.read_csv(file, usecols=["Unnamed: 0", "log_loss"])
    plt.plot(df["Unnamed: 0"], df["log_loss"], label=file)

    min_idx = df["log_loss"].idxmin()
#  print(f"{file}: min log_loss = {df['log_loss'].min()} at  {df.loc[min_idx, 'Unnamed: 0']}")

plt.xlabel("B")
plt.ylabel("OOS Log Loss")
plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left", borderaxespad=0.0)
plt.show()
```
